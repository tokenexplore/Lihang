logistic 分布

```python
import numpy as np
import matplotlib.pyplot as plt

def logistic_F(x,u,r):
    ep = np.exp(-(x-u)/r)
    return 1/(1+ep)

def logistic_f(x,u,r):
    ep = np.exp(-(x-u)/r)
    return ep/(r * np.square(1+ep))
```

```python
figure,axes = plt.subplots(2,5,figsize = (20,10))
axes = axes.flatten() # 将二维数组展平，便于遍历

for i in range(10):
    u = i-3
    r=1
    x = np.linspace(-10,10,1000)
    axes[i].plot(x,logistic_F(x,u,r),label=f'u={u},r={r}')
    axes[i].grid(True)
    axes[i].legend()
    
plt.show()
```

```python
figure,axes = plt.subplots(2,5,figsize = (20,10))
axes = axes.flatten() # 将二维数组展平，便于遍历

for i in range(10):
    u = i-10
    r=u^2-11
    x = np.linspace(-10,10,10000)
    axes[i].plot(x,logistic_f(x,u,r),label=f'u={u},r={r}')
    axes[i].grid(True)
    axes[i].legend()
    
plt.show()
```
